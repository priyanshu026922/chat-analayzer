```python
import re
import pandas as pd
```

```python
f=open('group-ChatHistory.txt','r',encoding='utf-8')
```

```python
data=f.read()
```

```python
data
```

```python
pattern='\d{1,2}/\d{1,2}/\d{2,4},\s\d{1,2}:\d{2}\s-\s'
```

```python
messages=re.split(pattern,data)[1:]#remove the  empty string at first
```

```python
dates=re.findall(pattern,data)
dates
```

```python
messages
```

```python
df = pd.DataFrame({'user_message': messages, 'message_date': dates})
    # convert message_date type
df['message_date'] = pd.to_datetime(df['message_date'].str.replace(' - ', ''), dayfirst=True, errors='coerce')

df.rename(columns={'message_date': 'date'}, inplace=True)
```

```python
df.shape
```

```python
df
```

```python
users=[]
messages=[]
for message in df['user_message']:
    entry=re.split('([\w\W]+?):\s',message)
    if entry[1:]:
        users.append(entry[1])
        messages.append(entry[2])
    else:
        users.append('Group_Notification')
        messages.append(entry[0])

df['user']=users
df['message']=messages
df.drop(columns=['user_message'],inplace=True)

df.head()
```
